## 3 · Five-fold cross-validation: give every day a turn

Here $k=5$. The 25 development days are in groups A–E of five days each. In round A, train on B–E and predict A. In the next round, start again at $(b,w)=(50,10)$ and hold out B. Rotate until every group has been held out once. Use exactly 200 updates and $\alpha=0.1$ in every fit.

For each fold, score its five predictions in kWh using mean absolute error, $\mathrm{MAE}=\frac{1}{5}\sum_i|y_i-\hat y_i|$. The overall score is the average of the five fold scores. Because the folds are the same size, this also equals the mean of all 25 held-out absolute errors.

```python
for letter, group in zip(letters, groups):
    print(f"Group {letter}:", ", ".join(f"D{i+1:02d}" for i in group))
print("Round A trains on B, C, D, E and validates on A.")
```

```python
def cross_validate(x, y, groups):
    all_ids = np.arange(len(y))
    visits = np.zeros(len(y), dtype=int)
    predictions = np.full(len(y), np.nan)
    fold_maes = []
    for held_out in groups:
        # TODO 5a. Keep the row indices outside held_out (use np.setdiff1d).
        train = None
        if train is None:
            return None
        assert len(np.intersect1d(train, held_out)) == 0
        assert len(train) + len(held_out) == len(y)
        # TODO 5b. Fit on the training rows. Use the default 200 steps.
        fitted = None
        if fitted is None:
            return None
        b, w, _ = fitted
        # TODO 5c. Apply that fold's fitted line to its held-out inputs.
        pred = None
        if pred is None:
            return None
        predictions[held_out] = pred
        visits[held_out] += 1
        fold_maes.append(np.mean(np.abs(y[held_out] - pred)))
    assert np.all(visits == 1) and np.isfinite(predictions).all()
    return np.asarray(fold_maes), predictions
```

```python
cv_ready = False
if fit_ready:
    result = cross_validate(x, energy, groups)
    if result is not None:
        fold_maes, held_preds = result
        assert len(fold_maes) == 5 and len(held_preds) == len(energy)
        # If round A never trains on A's meter readings, changing those readings
        # cannot change the predictions for A. It can change its score.
        changed = energy.copy()
        changed[groups[0]] += 100
        altered = cross_validate(x, changed, groups)
        np.testing.assert_allclose(held_preds[groups[0]], altered[1][groups[0]], atol=1e-8)
        pooled_mae = np.abs(energy - held_preds).mean()
        np.testing.assert_allclose(fold_maes.mean(), pooled_mae)
        cv_ready = True
        print("Held out   Trained on   Validated on   MAE (kWh)")
        for letter, group, score in zip(letters, groups, fold_maes):
            print(f"   {letter}           {len(x)-len(group):2d}              {len(group)}            {score:.3f}")
        print(f"Mean five-fold MAE: {fold_maes.mean():.3f} kWh")
        print("\nRound A | ID     observed   prediction   absolute error (kWh)")
        for i in groups[0]:
            print(f"          D{i+1:02d}    {energy[i]:7.2f}    {held_preds[i]:8.2f}       {abs(energy[i]-held_preds[i]):7.2f}")
if not cv_ready:
    print("Finish fit_line and cross_validate to see the five validation scores.")
```
