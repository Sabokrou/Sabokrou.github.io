## 6. Handle small counts with smoothing

### Understand the concept
Not observing an event does not prove that it is impossible. An unsmoothed zero feature likelihood can make an entire class score zero. Additive smoothing adds pseudo-counts for both possible binary outcomes. With alpha>0, estimate theta=(ones+alpha)/(class_count+2*alpha). The denominator has 2*alpha because the feature has two possible values, 0 and 1.

### Small worked example
If a feature is 1 zero times among eight class records, its unsmoothed probability is 0/8=0. With alpha=1 it becomes (0+1)/(8+2)=0.1. Its feature-zero probability is then 0.9. In our delivery table, late-class rain with alpha=1 becomes (6+1)/(8+2)=0.7.
